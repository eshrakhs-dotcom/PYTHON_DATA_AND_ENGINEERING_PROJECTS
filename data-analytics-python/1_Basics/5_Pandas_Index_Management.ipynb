```python
# Import the libraries needed for data analysis and visualization
import pandas as pd                              # Pandas: work with and analyze tabular data
from datasets import load_dataset               # Hugging Face: load the job postings dataset
import matplotlib.pyplot as plt                 # Matplotlib: create charts and visualizations

# Load the dataset and convert it into a pandas DataFrame
# This gives us a table-like structure for analysis
dataset = load_dataset('lukebarousse/data_jobs') # Download/load Luke's job dataset
df = dataset['train'].to_pandas()                # Convert the dataset into a Pandas DataFrame

# Clean up the date column by converting it to datetime values
# This makes date-based analysis easier and more accurate
df['job_posted_date'] = pd.to_datetime(
    df['job_posted_date']
)  
```

```python
# Show a small sample of rows from the DataFrame
# This helps us quickly inspect the data without displaying everything

df.sample(5)
```

```python
# Display the current index values of the DataFrame
# The index is the row labels used by pandas to identify each row

df.index
```

```python
# Give the DataFrame index a descriptive name
# This makes the index easier to understand when displayed or used later

df.index.name = 'job_index'
```

```python
# Check the data type of the index
# This shows whether the index is numeric, object-based, or another pandas dtype

df.index.dtype
```

## Three methods to cover

- df.reset_index()
- df.set_index()
- df.sort-index()

```python
# Filter the DataFrame to only include rows from the United States
# This uses boolean indexing to keep matching rows

df_usa = df[df['job_country'] == 'United States']
```

## Set Index


```python
# Reset the index of the filtered DataFrame
# inplace=True applies the change directly to df_usa instead of creating a copy

df_usa.reset_index(inplace=True)

# Display the updated DataFrame
# The old index values are now inserted as a new column

df_usa
```

## Sort Index


```python
# Create a pivot table summarizing annual salary by job title
median_pivot = df_usa.pivot_table(
    values='salary_year_avg',                 # Values to summarize: annual salary
    index='job_title_short',                  # Rows: each job title becomes an index label
    aggfunc=['median', 'min', 'max']          # Calculate median, minimum, and maximum salary
)

# Sort the pivot table by its index (job titles) alphabetically
median_pivot = median_pivot.sort_index()

# Display the result
median_pivot
```
