```python
dataset = load_dataset('lukebarousse/data_jobs') # Download/load Luke's job dataset
df_original = dataset['train'].to_pandas()
```

```python
df_altered = df_original.copy()
```

```python
# ---------------------------------------------------------
# CHECKING IF TWO DATAFRAMES ARE SEPARATE OBJECTS
# ---------------------------------------------------------

# id() returns the unique memory identity of a Python object
print('ID of df_original:              ', id(df_original))
print('ID of df_altered:               ', id(df_altered))

# Compare their IDs
# False means df_original and df_altered are separate DataFrame objects
print('Are the two dataframes the same?', id(df_original) == id(df_altered))


# ---------------------------------------------------------
# FILLING MISSING SALARY DATA
# ---------------------------------------------------------

# Calculate the median (middle) annual salary
# Pandas automatically ignores NaN/missing values when calculating median
median_salary = df_altered['salary_year_avg'].median()

# Replace missing (NaN) salary values with the median salary
# The cleaned values are assigned back into the salary_year_avg column
df_altered['salary_year_avg'] = df_altered['salary_year_avg'].fillna(median_salary)

# Display the first 5 rows of the salary column to check the result
df_altered.loc[:5, 'salary_year_avg']
```
